### 6. API Configuration
Define the base endpoint URLs and parameter structures for the Open-Meteo API.

````python
API_URL = "https://api.open-meteo.com/v1/forecast"
DEFAULT_CURRENT_PARAMS = "temperature_2m,relative_humidity_2m,wind_speed_10m"
````

### 7. API Function
Function responsible for fetching weather data from the API endpoint.

````python
def fetch_weather_raw(latitude, longitude):
    params = {
        "latitude": latitude,
        "longitude": longitude,
        "current": DEFAULT_CURRENT_PARAMS
    }
    response = requests.get(API_URL, params=params, timeout=10)
    response.raise_for_status()
    return response.json()
````

### 8. Data Processing
Process raw JSON API responses and parse them into structured structures.

````python
def process_weather_data(raw_data):
    if not raw_data or "current" not in raw_data:
         raise ValueError("Invalid weather data structure received")

    current = raw_data["current"]
    return {
        "temperature": current["temperature_2m"],
        "humidity": current["relative_humidity_2m"],
        "wind_speed": current["wind_speed_10m"]
    }
````

### 9. Data Display
Display the structured weather metrics in a professional terminal interface.

````python
def display_weather(processed_data):
    print("===============================")
    print("       WEATHER INFORMATION     ")
    print("===============================")
    print(f"Temperature : {processed_data['temperature']} °C")
    print(f"Humidity    : {processed_data['humidity']} %")
    print(f"Wind Speed  : {processed_data['wind_speed']} km/h")
````

### 10. User Input
Gather latitude and longitude parameters safely from the user interface.

````python
def get_user_coordinates():
    try:
        lat = float(input("Enter latitude: "))
        lon = float(input("Enter longitude: "))
        return lat, lon
    except ValueError:
        print("Invalid numeric input! Please enter floating point coordinates.")
        return None
````

### 11. Error Handling
Wrap the orchestration and integration flow in robust connection, API, and validation error handles.

````python
def run_weather_app():
    coords = get_user_coordinates()
    if not coords:
        return

    lat, lon = coords
    try:
        raw_data = fetch_weather_raw(lat, lon)
        processed_data = process_weather_data(raw_data)
        display_weather(processed_data)
    except requests.exceptions.RequestException as e:
        print("\nError: Unable to connect to the weather service.")
        print("Details:", e)
    except (ValueError, KeyError) as e:
        print("\nError: Failed to process weather details structure.")
        print("Details:", e)
````

### 12. Testing
Execute the weather flow program in real time.
